#### What is Numpy
Numpy is one of the fundamental Python packages for data science because of its ability to manipulate numerical data. Many other packages in the PyData ecosystem are built upon Numpy, including Pandas. Numpy is shorthand for "numerical Python."

#### What is Pandas
Pandas is one of the most popular data manipulation and analysis libraries. It provides intuitive, high-performance data structures—like DataFrames (tables) and Series (lists)—that make cleaning, transforming, and analyzing large datasets feel just like working in Excel.

*You need to install Numpy and Pandas before moving on.  They should be included with your Anaconda installation.*

#### 5. Data structures
A data structure is a specialized format for organizing, managing, and storing data so it can be accessed and modified efficiently. We will primarily focus on (1) lists, (2) series, and (3) dataframes.

##### 5.1 Lists
A list is a built-in data structure used to store an ordered, mutable (changeable) collection of items inside square brackets [], with each item separated by a comma.
Items DO NOT have to be of the same data type.

##### 5.2 Series
A series is a one-dimensional data structure provided by Pandas. It can hold data of any data type, such as int, str (denoted object), and float. The series has a few advantages over lists: it is extremely fast and memory-efficient, can handle missing values, and better supports math operations.

##### 5.3 DataFrames
A Python DataFrame is a two-dimensional, tabular data structure that functions similarly to an Excel spreadsheet/table.
