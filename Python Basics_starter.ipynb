#### 5. Data structures
A data structure is a specialized format for organizing, managing, and storing data so it can be accessed and modified efficiently. We will primarily focus on (1) lists, (2) series, and (3) dataframes.

##### 5.2 Series
A series is a one-dimensional data structure provided by Pandas. It can hold data of any data type, such as int, str (denoted object), and float. The series has a few advantages over lists: it is extremely fast and memory-efficient, can handle missing values, and better supports math operations.

```python
a1 = pd.Series([1, 2, 3, 4])
a1 # print result is different from lists, bc there is an index on the left and the datatype below
```

```python
b1 = pd.Series(['a', 'b', 'c', 'd'])
b1 # data type is object, which a broader term for strings
```

```python
c1 = pd.Series([1, "b", 3, "d"])
c1 # data type is object bc strings are a part of it
```

```python
a2 = a1 + 2 # adds two to each item in the series
a2
```

```python
numbers2 = numbers + 2 # this will not work bc you can't add a number to a list, but you can add a number to a series
```

##### 5.3 DataFrames
A Python DataFrame is a two-dimensional, tabular data structure that functions similarly to an Excel spreadsheet/table.

```python
# creating a dataframe using pandas library, the pd.DataFrame() function must be typed like that
df = pd.DataFrame({
    'id':['1', '2', '3'], # in class we won't have to type, we'll rly j import data as a dataframe
    'price':[10, 5, 8]
})
df 
```

```python
# say the prices all went up by 1, we can add 1 to the price column
df['price'] + 1 # square bracket selects a column within dataframe "df"
```

```python
df # see how it didn't change the original dataframe and just printed the new values?
```

```python
df['price'] = df['price'] + 1 # this will change the original dataframe, bc we are assigning the new values to the column
df
```

#### 6. Methods
A method is a function that is explicitly tied to an object or class. We will apply the following methods to DataFrames:

- .copy()
- .info()
- .head()
- .tail()
- .describe()

*[these are specific to pandas]*

```python
# copy() creates a copy of the dataframe. Note how its not like excel where you have to have an argument, this function will j go
df2 = df.copy()
df2
```

```python
df3 = pd.DataFrame({
    'id':['1', '2', '3', '4'],
    'price':[10, 5, 8, np.nan] # nan here means missing value (like null), it's from numpy library saved as np
})
df3
```

```python
# info() gives basic summary of the dataframe' structure
df3.info()  # the dtype for the price column is float64 bc it has a missing value, which is a float
```

```python
# head() and tail() show first few and last few rows of the dataframe
df3.head(3) # by default with no argument, it's 5 rows, but you can specify with a number
```

```python
df3.tail(2) # last two rows (5 default again)
```

```python
# 2 ways to use:
df3.describe()  # gives summary statistics for all numerical columns
```

```python
df3['price'].describe() # gives summary statistics for just the price column
```
